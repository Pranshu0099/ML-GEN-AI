***Question 1: What is Deep Learning? Briefly describe how it evolved and how it differs from traditional machine learning.***

**Answer:** Deep Learning (DL) is a specialized subset of Machine Learning (ML) that uses multi-layered Artificial Neural Networks (ANNs) to automatically extract higher-level features and learn intricate representations directly from raw data.

> Architecture: It consists of an input layer, multiple hidden layers (hence the term "deep"), and an output layer.

> Mechanism: Using non-linear transformations across successive layers, deep networks learn complex patterns without relying on human intervention to specify feature extractors.

**Evolution of Deep Learning**

The development of deep learning evolved through three major historical phases:

1. The Foundation & Early Perceptron Era (1950s–1960s):

    1958: Frank Rosenblatt invented the Perceptron, the simplest form of a single-layer neural network.

    Stagnation (First AI Winter): In 1969, Marvin Minsky and Seymour Papert mathematically proved that a single-layer Perceptron could not solve simple non-linear problems like the XOR operation, halting neural network research for years.

2. lgorithm Refinement & Multi-Layer Networks (1980s–1990s):

    1986: Rumelhart, Hinton, and Williams popularized the Backpropagation algorithm, enabling effective training of multi-layer neural networks using gradient descent.

    1989/1998: Yann LeCun developed Convolutional Neural Networks (CNNs) (LeNet-5) for computer vision tasks like handwritten digit recognition.

    Constraint: Progress remained limited due to small training datasets and insufficient computing power (CPUs).

3. The Modern Deep Learning Boom (2010s–Present):

    2012 (AlexNet Breakthrough): Alex Krizhevsky, Ilya Sutskever, and Geoffrey Hinton won the ImageNet competition using a deep CNN (AlexNet) trained on Graphics Processing Units (GPUs), dramatically outperforming traditional computer vision techniques.

**Key Differences: Traditional Machine Learning vs. Deep Learning**

1. Feature Extraction:

    Traditional Machine Learning: Relies heavily on manual feature engineering, requiring domain experts to hand-craft and select relevant features (e.g., HOG or SIFT filters) before feeding them into the algorithm.

    Deep Learning: Performs automated feature extraction, where the network independently discovers and learns hierarchical features directly from raw, unprocessed data.

2. Data Dependency and Scaling:

    Traditional Machine Learning: Performs well on small to medium-sized datasets, but its performance tends to plateau relatively quickly as more data is added.

    Deep Learning: Requires massive volumes of data to generalize effectively; performance continues to scale and improve significantly as data size increases, though it risks severe overfitting if data is insufficient.

3. Computational Hardware Requirements:

    Traditional Machine Learning: Can be trained and executed efficiently on standard Central Processing Units (CPUs) with modest hardware requirements.

    Deep Learning: Heavily dependent on specialized high-performance parallel computing hardware, primarily Graphics Processing Units (GPUs) and Tensor Processing Units (TPUs), to handle matrix operations.

4. Model Interpretability:

    Traditional Machine Learning: Generally offers higher transparency and interpretability ("White-box" models like Decision Trees, Linear Regression, and Support Vector Machines) where decision logic can be easily traced.

    Internal Transparency (Deep Learning): Often operates as a "Black-box" model, where millions of interconnected parameters make it extremely difficult to trace the exact chain of reasoning behind a specific output.

5. Primary Data Types:

    Traditional Machine Learning: Most effective when applied to structured and tabular data (e.g., relational databases, CSV spreadsheets, discrete numerical/categorical columns).

    Deep Learning: Excels exceptionally at handling unstructured, high-dimensional data (e.g., raw images, video streams, audio waveforms, and natural language text).

6. Training Time and Complexity:

    Traditional Machine Learning: Features fast training times ranging from seconds to a few hours, with straightforward hyperparameter tuning.

    Deep Learning: Involves complex architecture design and extensive training cycles that can take anywhere from hours to weeks, requiring substantial computational infrastructure.

**Question 2: Explain the basic architecture and functioning of a Perceptron. What are its limitations?**

**Answer:**

**Basic Architecture of a Perceptron**

Introduced by Frank Rosenblatt in 1958, the Perceptron is the foundational building block of artificial neural networks. It is a single-layer binary linear classifier that maps inputs to a binary output value.

The architecture consists of Inputs ($x_1, x_2, \dots, x_n$), Weights ($w_1, w_2, \dots, w_n$), Bias ($b$), Net Input Function (Weighted Sum), Activation Function.

**Functioning of a Perceptron**

The functioning of a Perceptron can be broken down into a two-step mathematical process followed by an iterative learning algorithm:

1. Forward Propagation (Prediction):Each input $x_i$ is multiplied by its corresponding weight $w_i$.All weighted products are summed together, and the bias $b$ is added to form the net input $z$.The activation function evaluates $z$. If $z$ is greater than or equal to a chosen threshold (often $0$), the activation function outputs $1$; otherwise, it outputs $0$ (or $-1$).

2. The Perceptron Learning Rule (Training):The model starts with random initial weights and bias values.For each training sample, it makes a prediction ($\hat{y}$) and compares it to the true target label ($y$).If the prediction is correct, no changes are made.If the prediction is incorrect, the weights and bias are updated using the Perceptron Learning Rule to minimize the error.

**Limitations of a Perceptron**

1. Inability to Solve Non-Linear Problems (The XOR Problem):

    A single Perceptron can only draw a linear decision boundary (a straight line in 2D space, a plane in 3D, or a hyperplane in higher dimensions).

2. Binary Classification Constraint:

    The traditional Perceptron is strictly designed for binary classification tasks and cannot directly output continuous probability values or handle multi-class classification without structural modifications (e.g., softmax layers or multi-class architectures).

3. Sensitivity to Outliers and Initialization:

    Because the learning rule updates weights exclusively based on misclassified points, the final decision boundary can be heavily influenced by noisy data or outliers, and convergence depends heavily on the initial random weights and learning rate selection.

**Question 3: Describe the purpose of activation function in neural networks. Compare Sigmoid, ReLU, and Tanh functions.**

**Answer:**

**Purpose of Activation Functions in Neural Networks**

An activation function is a mathematical gate applied to the output of a neural network's weighted sum of inputs before passing it to the next layer. Its primary purposes are:

1. Introducing Non-Linearity:
    Without activation functions, a neural network—regardless of how many hidden layers it contains—would function merely as a giant single-layer linear regressor (since linear combinations of linear functions remain linear). Activation functions allow networks to learn complex, non-linear decision boundaries, mapping intricate patterns in data (such as images, speech, and text).

2. Controlling Signal Propagation (Firing Mechanism):
    Inspired by biological neurons that fire only when a certain threshold of electrical potential is reached, activation functions determine whether a specific neuron should be "activated" (pass its signal forward) or suppressed.

3. Enabling Gradient-Based Optimization:
    Because activation functions are differentiable, they allow the network to compute gradients via backpropagation, enabling the optimization of weights and biases through gradient descent.

**Comparison of Sigmoid, ReLU, and Tanh Functions**

**1. The Sigmoid Function**

Mathematical Formula:$$\sigma(x) = \frac{1}{1 + e^{-x}}$$Output Range: $(0, 1)$


Characteristics & Use Cases:Maps any real-valued number into a probability-like range between $0$ and $1$, making it historically popular for the output layer of binary classification models.

Advantages:  

Smooth gradient, preventing abrupt jumps in output values.  
Clear interpretation as a probability or confidence score.

Limitations:

Vanishing Gradient Problem: For extremely high or low input values, the derivative of the sigmoid function approaches zero, halting weight updates during backpropagation.

Not Zero-Centered: Outputs are strictly positive, which can cause gradient updates to oscillate in inefficient directions during optimization.

Computationally Expensive: Involves exponential calculations.


**2. The Tanh (Hyperbolic Tangent) **

FunctionMathematical Formula:$$\tanh(x) = \frac{e^x - e^{-x}}{e^x + e^{-x}}$$Output Range: $(-1, 1)$

Characteristics & Use Cases: Rescaled version of the sigmoid function, mapping inputs to a symmetric range around zero. Commonly used in hidden layers of recurrent neural networks (RNNs).

Advantages:

Zero-Centered: Because outputs range from $-1$ to $+1$, the mean of the activations across hidden layers is closer to zero, making optimization smoother and faster compared to the sigmoid function.

Steeper gradients than the sigmoid function, improving learning dynamics.


Limitations:

Vanishing Gradient Problem: Like sigmoid, it still suffers from vanishing gradients for very large positive or negative inputs because its derivative flattens out at the extremes.

Computationally expensive due to exponential operations.


**3. ReLU (Rectified Linear Unit) Function**

Mathematical Formula:$$f(x) = \max(0, x)$$Output Range: $[0, \infty)$ (Zero for any negative input, linear identity for any positive input).

Characteristics & Use Cases:The default activation function for modern deep hidden layers, especially in Convolutional Neural Networks (CNNs) and Deep Neural Networks (DNNs).

Advantages:

Computational Efficiency: Involves simple thresholding (comparing to zero), requiring no expensive exponential or division operations.

Mitigates Vanishing Gradient: For all positive inputs, the gradient is constant ($1.0$), allowing deep networks to train efficiently without gradients shrinking exponentially across layers.

Sparse Activation: Outputs exact zeros for negative inputs, meaning only a subset of neurons activate at any given time, creating a more efficient network representation.

Limitations:

The "Dying ReLU" Problem: If a large gradient flows through a ReLU neuron, it can update weights such that the neuron consistently outputs zero for all subsequent inputs. Once a neuron's activation becomes permanently zero, its gradient becomes zero, and it stops learning permanently.

**Question 4: What is the difference between Loss function and Cost function in neural
networks? Provide examples.**

**Answer:**

**Definitions and Core Distinctions**

While the terms Loss Function and Cost Function are frequently used interchangeably in literature and practice, they have a precise technical distinction based on their scope of computation:

>Loss Function:A loss function is evaluated on a single training example or data point. It computes the numerical error or penalty between the model's predicted output ($\hat{y}$) and the true target value ($y$) for that specific instance.

Formula context: $L(y, \hat{y})$


>Cost Function (Objective Function):A cost function is the average or aggregated sum of loss functions computed across the entire training dataset (or a mini-batch). It provides a global measure of how well the neural network is performing overall.

Formula context: $J(w, b) = \frac{1}{N} \sum_{i=1}^{N} L(y_i, \hat{y}_i)$ (where $N$ is the total number of samples).

**Key Differences**

1. Scope of Application:

A loss function operates at the micro-level, targeting individual data points to analyze specific errors.

A cost function operates at the macro-level, summarizing performance across the entire dataset to guide global optimization.

2. Role in Optimization:

The loss provides the feedback signal for a single instance during forward and backward passes.

The cost value is the scalar metric minimized by gradient descent algorithms (e.g., updating weights $w$ and biases $b$) to improve the network's overall predictive capability.

3. Terminology Usage:

In some frameworks (like PyTorch or TensorFlow), a "loss function" object often computes both the individual losses and averages them over a batch, but conceptually the underlying mathematics separates the instance-level penalty from the dataset-wide objective.

**Examples of Loss and Cost Functions**

1. Mean Squared Error (MSE) / Quadratic Loss

Loss Function (Single Sample):$$L(y, \hat{y}) = (y - \hat{y})^2$$Explanation: Measures the squared difference between the true target $y$ and the prediction $\hat{y}$ for one specific row of data.

Cost Function (Entire Dataset):$$J(w, b) = \frac{1}{N} \sum_{i=1}^{N} (y_i - \hat{y}_i)^2$$Explanation: The average of the squared errors across all $N$ samples in the dataset, commonly used in regression tasks.

2. Binary Cross-Entropy Loss

Loss Function (Single Sample):$$L(y, \hat{y}) = - \left[ y \log(\hat{y}) + (1 - y) \log(1 - \hat{y}) \right]$$

Explanation: Calculates the penalty for a single binary classification prediction, where $y \in \{0, 1\}$ and $\hat{y}$ is the predicted probability.

Cost Function (Entire Dataset):$$J(w, b) = - \frac{1}{N} \sum_{i=1}^{N} \left[ y_i \log(\hat{y}_i) + (1 - y_i) \log(1 - \hat{y}_i) \right]$$

Explanation: The mean cross-entropy loss across all training samples, used to optimize binary classification neural networks.

**Question 5: What is the role of optimizers in neural networks? Compare Gradient Descent, Adam, and RMSprop.**

**Answer:**
**The Role of Optimizers in Neural Networks**

An optimizer is a mathematical algorithm or method used to update the weights and biases of a neural network during training, with the ultimate goal of minimizing the cost function (i.e., reducing the error between predictions and true labels).

    The Optimization Process: After backpropagation computes the gradient of the cost function with respect to each parameter, the optimizer determines how and by how much those parameters should be adjusted.

    Guiding the Descent: It navigates the high-dimensional loss landscape, helping the network escape inefficient regions (such as flat areas, saddle points, or local minima) to find the global (or a near-optimal) minimum efficiently.

**Comparison of Gradient Descent, Adam, and RMSprop**

1. Gradient Descent (Standard / SGD / Mini-batch)Mechanism:

Gradient Descent computes the gradient of the cost function across the dataset (or subsets in mini-batch SGD) and updates parameters directly in the opposite direction of the gradient, scaled by a fixed learning rate $\alpha$:$$w_{t+1} = w_t - \alpha \nabla J(w_t)$$

Advantages:

Simple to understand and implement.

Guaranteed convergence to the global minimum for convex error surfaces.

Limitations:

Single Learning Rate: Uses the same learning rate for all parameters, which causes issues if some features are sparse or occur much more frequently than others.

Navigation Issues: Struggles in ravines or areas where the surface curves steeply in one dimension and gently in another, leading to slow oscillation and delayed convergence.

Saddle Points: Easily gets trapped in saddle points or shallow local minima where gradients approach zero.


2. RMSprop (Root Mean Square Propagation)Mechanism:
Developed to resolve the radical oscillations of standard gradient descent, RMSprop adapts the learning rate for each individual parameter. It divides the learning rate by an exponentially decaying average of squared gradients ($v_t$):$$v_t = \beta v_{t-1} + (1 - \beta)(\nabla J(w_t))^2$$$$w_{t+1} = w_t - \frac{\alpha}{\sqrt{v_t} + \epsilon} \odot \nabla J(w_t)$$(where $\beta$ is the moving average parameter and $\epsilon$ prevents division by zero).

Advantages:

Adaptive Learning Rates: Automatically scales the step size based on recent gradient magnitudes, dampening oscillations along steep dimensions and accelerating progress along gentle slopes.

Highly effective for non-stationary settings and Recurrent Neural Networks (RNNs).

Limitations:

Still relies on a global initial learning rate that must be manually tuned.

The running average of squared gradients can cause the effective learning rate to shrink too aggressively over long training runs.

3. Adam (Adaptive Moment Estimation)Mechanism:
Adam combines the principles of both Momentum (storing a running average of past gradients—the first moment) and RMSprop (storing a running average of past squared gradients—the second moment), incorporating bias-correction terms for both moments:$$m_t = \beta_1 m_{t-1} + (1 - \beta_1)\nabla J(w_t) \quad \text{(First Moment / Momentum)}$$$$v_t = \beta_2 v_{t-1} + (1 - \beta_2)(\nabla J(w_t))^2 \quad \text{(Second Moment / Uncentered Variance)}$$$$w_{t+1} = w_t - \frac{\alpha}{\sqrt{\hat{v}_t} + \epsilon} \hat{m}_t$$

Advantages:

Fast and Reliable Convergence: Combines momentum's ability to barrel through ravines with RMSprop's per-parameter adaptive learning rates.

Robust to Hyperparameters: Generally requires minimal tuning of its default hyperparameters ($\alpha = 0.001$, $\beta_1 = 0.9$, $\beta_2 = 0.999$), making it the de facto standard default optimizer for most deep learning architectures.Handles sparse gradients and noisy data exceptionally well.

Limitations:

In some specific computer vision or fine-tuning tasks, tightly tuned Stochastic Gradient Descent with Momentum (SGD+Momentum) can achieve better generalization performance than Adam, though Adam typically trains faster.

```python
'''
Question 6: Write a Python program to implement a single-layer perceptron from
scratch using NumPy to solve the logical AND gate
'''

import numpy as np

# training data as NumPy arrays
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

y = np.array([0, 0, 0, 1])

# 2. Initialize weights and bias to zeros
# We have 2 input features, so we create an array of 2 zeros
weights = np.array([0.0, 0.0])
bias = 0.0

# 3. Setting hyperparameters
learning_rate = 0.1
epochs = 10

print("--- Training Started ---\n")

# 4. Training loop over epochs
for epoch in range(epochs):
    print(f"Epoch {epoch + 1} of {epochs}")

    # Loop through each sample in our dataset
    for i in range(len(X)):
        xi = X[i]          # Input features [x1, x2]
        target = y[i]      # Correct answer

        # Step A: Calculate weighted sum using NumPy dot product (w1*x1 + w2*x2)
        linear_output = np.dot(xi, weights) + bias

        # Step B: Activation function (Step function)
        prediction = 1 if linear_output >= 0 else 0

        # Step C: Error calculation
        error = target - prediction

        # Step D: Update weights and bias if prediction is wrong
        if error != 0:
            weights += learning_rate * error * xi
            bias += learning_rate * error

        print(f"  Input: {xi} | Target: {target} | Predicted: {prediction} | Error: {error}")
    print("-" * 40)

print("\n--- Training Finished ---")
print(f"Final Weights: {weights}")
print(f"Final Bias:    {bias}\n")

# Testing

print("--- Testing the Trained Perceptron ---")

# Calculate predictions for all inputs at once using NumPy dot product
all_outputs = np.dot(X, weights) + bias
predictions = np.where(all_outputs >= 0, 1, 0)

for i in range(len(X)):
    print(f"Input: {X[i]} | Predicted Output: {predictions[i]} | Correct Target: {y[i]}")
```

```python
'''
Question 8: Use Keras to build and train a simple multilayer neural network on the
MNIST digits dataset. Print the training accuracy.
'''

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten


# 1. Load and Preprocess the MNIST Dataset

print("Loading MNIST dataset...")
mnist = tf.keras.datasets.mnist
(X_train, y_train), (X_test, y_test) = mnist.load_data()

# Normalize pixel values from [0, 255] to [0.0, 1.0] for stable training
X_train = X_train / 255.0
X_test = X_test / 255.0


# 2. Build the Multilayer Neural Network

model = Sequential([
    # Input layer: flattens 28x28 pixel grid into 784 features
    Flatten(input_shape=(28, 28)),

    # Hidden layer: 128 neurons with ReLU activation
    Dense(128, activation='relu'),

    # Output layer: 10 neurons (digits 0-9) with Softmax activation for multi-class classification
    Dense(10, activation='softmax')
])


# 3. Compile the Model

model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)


# 4. Train the Model

print("\nTraining the model...")
history = model.fit(
    X_train,
    y_train,
    epochs=5,          # Number of complete passes over the training data
    batch_size=32,     # Number of samples per gradient update
    verbose=1          # Show training progress bar
)


# 5. Extract and Print Training Accuracy

# Retrieve accuracy from the final training epoch
final_train_accuracy = history.history['accuracy'][-1] * 100

print("\n--- Training Results ---")
print(f"Final Training Accuracy: {final_train_accuracy:.2f}%")
```

**Question 10: You are working on a project for a bank that wants to automatically detect fraudulent transactions. The dataset is large, imbalanced, and contains structured features like transaction amount, merchant ID, and customer location. The goal is to classify each transaction as fraudulent or legitimate.**

**Explain your real-time data science workflow:
● How would you design a deep learning model (perceptron or multilayer NN)?
● Which activation function and loss function would you use, and why?
● How would you train and evaluate the model, considering class imbalance?
● Which optimizer would be suitable, and how would you prevent overfitting?**

**Answer:**

**Part A: Deep Learning Model Design**

For a tabular fraud detection dataset with structured features (transaction amount, merchant ID, customer location, etc.), a Multi-Layer Perceptron (MLP)—also known as a Feedforward Neural Network—is the appropriate deep learning architecture. While tree-based ensemble models (such as XGBoost or LightGBM) often dominate tabular benchmarks, an MLP can effectively learn complex, non-linear feature interactions when properly scaled and regularized.

Input Layer: Receives the preprocessed feature vector. Numerical features (e.g., transaction amount) are standardized, while high-cardinality categorical features (e.g., merchant ID, location) are handled via entity embeddings or one-hot encoding.

Hidden Layers: A sequence of dense (fully connected) layers with decreasing widths (e.g., 128 $\to$ 64 $\to$ 32 neurons) to progressively extract higher-level representations of suspicious transaction patterns.

Output Layer: A single neuron with a Sigmoid activation function to output a continuous probability score between $0$ (legitimate) and $1$ (fraudulent).

**Part B: Activation and Loss Functions**

Activation Functions:

Hidden Layers — ReLU (Rectified Linear Unit): Used across all hidden layers ($f(x) = \max(0, x)$) because it introduces non-linearity efficiently, avoids the vanishing gradient problem during deep backpropagation, and computes rapidly.

Output Layer — Sigmoid Function: $\sigma(x) = \frac{1}{1 + e^{-x}}$. Essential for binary classification as it maps the final logit output to a bounded probability range of $(0, 1)$.

Loss Function:

Binary Cross-Entropy Loss (BCE):$$L(y, \hat{y}) = - \left[ y \log(\hat{y}) + (1 - y) \log(1 - \hat{y}) \right]$$Why: It penalizes confident misclassifications heavily, making it mathematically optimal for probability-based binary outputs.

**Part C: Training, Evaluation, and Handling Class Imbalance**

Fraud datasets are heavily imbalanced (fraudulent transactions often constitute less than $0.1\%$ of the data). Standard accuracy is misleading; a naive model predicting all transactions as legitimate achieves $>99\%$ accuracy while failing completely.

1. Handling Class Imbalance:

Resampling Techniques: Apply SMOTE (Synthetic Minority Over-sampling Technique) or class weighting during training to penalize misclassifications of the minority class more severely.

Stratified K-Fold Cross-Validation: Ensures that every training and validation fold maintains the exact same proportion of minority class samples as the overall dataset.

2. Evaluation Metrics:

Precision-Recall AUC (PR-AUC): The primary metric because it focuses heavily on the minority (positive) class.

F1-Score & Confusion Matrix: To evaluate the harmonic mean of precision and recall at an optimized decision threshold.

Avoid: Standard ROC-AUC can be overly optimistic on severely imbalanced datasets.

**Part D: Optimizer and Regularization (Preventing Overfitting)**

Optimizer:

Adam (Adaptive Moment Estimation): Combines momentum and RMSprop, providing adaptive learning rates per parameter. It handles sparse gradients and noisy data efficiently, making it ideal for financial fraud landscapes.

Preventing Overfitting:

Dropout Layers: Randomly deactivate a fraction of neurons (e.g., $p = 0.3$) during each training step to prevent co-adaptation of features.

L2 Regularization (Weight Decay): Penalizes excessively large weight coefficients in the loss function.

Early Stopping: Monitors validation loss and halts training automatically if validation performance fails to improve for a set number of epochs (patience).



```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import classification_report, average_precision_score, precision_recall_curve


# 1. Simulate Imbalanced Tabular Fraud Data

np.random.seed(42)
tf.random.set_seed(42)

print("Generating simulated tabular transaction data...")
num_samples = 20000
num_features = 20

# Generate normal transactions (99%) and fraudulent transactions (1%)
X_data = np.random.randn(num_samples, num_features)
y_data = np.random.choice([0, 1], size=num_samples, p=[0.99, 0.01])

# Split into train, validation, and test sets (Stratified)
X_train, X_test, y_train, y_test = train_test_split(
    X_data, y_data, test_size=0.2, random_state=42, stratify=y_data
)
X_train, X_val, y_train, y_val = train_test_split(
    X_train, y_train, test_size=0.2, random_state=42, stratify=y_train
)

# Feature Scaling (Crucial for Neural Networks on Tabular Data)
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_val = scaler.transform(X_val)
X_test = scaler.transform(X_test)

# Calculate class weights to handle severe imbalance
neg_count = np.sum(y_train == 0)
pos_count = np.sum(y_train == 1)
total_count = len(y_train)

weight_for_0 = (1.0 / neg_count) * (total_count / 2.0)
weight_for_1 = (1.0 / pos_count) * (total_count / 2.0)
class_weights = {0: weight_for_0, 1: weight_for_1}

print(f"Training samples: {total_count} | Fraud cases: {pos_count}")


# 2. Build the MLP Fraud Detection Model

model = Sequential([
    Dense(64, activation='relu', input_shape=(num_features,)),
    Dropout(0.3),  # Prevent overfitting
    Dense(32, activation='relu'),
    Dropout(0.3),  # Prevent overfitting
    Dense(1, activation='sigmoid') # Binary classification output
])

model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy', tf.keras.metrics.Precision(name='precision'), tf.keras.metrics.Recall(name='recall')]
)


# 3. Train Model with Regularization & Callbacks

# Early stopping to halt training if validation loss stops improving
early_stopping = EarlyStopping(
    monitor='val_loss',
    patience=5,
    restore_best_weights=True
)

print("\nTraining MLP model for fraud detection...")
history = model.fit(
    X_train, y_train,
    validation_data=(X_val, y_val),
    epochs=30,
    batch_size=64,
    class_weight=class_weights,  # Handle class imbalance
    callbacks=[early_stopping],
    verbose=1
)


# 4. Evaluate Model Performance

print("\n--- Evaluating Model on Test Set ---")
y_pred_probs = model.predict(X_test).ravel()

# Convert probabilities to binary predictions using standard 0.5 threshold (or tuned threshold)
y_pred_classes = (y_pred_probs >= 0.5).astype(int)

# Compute PR-AUC (best metric for imbalanced fraud data)
pr_auc = average_precision_score(y_test, y_pred_probs)
print(f"Precision-Recall AUC (PR-AUC): {pr_auc:.4f}")

# Print classification report
print("\nClassification Report:")
print(classification_report(y_test, y_pred_classes, target_names=['Legitimate (0)', 'Fraud (1)']))
```
